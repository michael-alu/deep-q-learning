```python
MEMBER = "michael"

EXPERIMENTS = {
    "michael": [
        "--learning-rate 0.0001 --gamma 0.99",
        "--learning-rate 0.0005 --gamma 0.99",
        "--learning-rate 0.001 --gamma 0.99",
        "--learning-rate 0.00005 --gamma 0.99",
        "--learning-rate 0.00025 --gamma 0.99",
        "--learning-rate 0.0001 --gamma 0.95",
        "--learning-rate 0.0001 --gamma 0.9",
        "--learning-rate 0.0001 --gamma 0.999",
        "--learning-rate 0.00025 --gamma 0.95",
        "--learning-rate 0.00025 --gamma 0.995",
    ],
    "merveille": [
        "--batch-size 16",
        "--batch-size 64",
        "--batch-size 128",
        "--batch-size 256",
        "--batch-size 16 --learning-rate 0.00005",
        "--batch-size 64 --learning-rate 0.00005",
        "--batch-size 64 --learning-rate 0.00025",
        "--batch-size 128 --learning-rate 0.00025",
        "--batch-size 128 --learning-rate 0.0005",
        "--batch-size 256 --learning-rate 0.0005",
    ],
    "ivan": [
        "--epsilon-decay-steps 50000",
        "--epsilon-decay-steps 200000",
        "--epsilon-decay-steps 300000",
        "--epsilon-end 0.01",
        "--epsilon-end 0.1",
        "--epsilon-end 0.2",
        "--epsilon-start 0.5",
        "--epsilon-start 0.5 --epsilon-decay-steps 50000",
        "--epsilon-end 0.01 --epsilon-decay-steps 50000",
        "--epsilon-start 0.8 --epsilon-end 0.02 --epsilon-decay-steps 150000",
    ],
}
```

```python
import shutil
from pathlib import Path

if shutil.which("uv") is None:
    raise RuntimeError("uv was not found. Re-run the setup cell above and check that it printed a uv version.")

output_folder = Path(f"/content/drive/MyDrive/dqn-experiments/{MEMBER}")

output_folder.mkdir(parents=True, exist_ok=True)

for experiment_number, settings in enumerate(EXPERIMENTS[MEMBER], start=1):
    model_path = output_folder / f"experiment_{experiment_number}.pth"

    if model_path.exists():
        print(f"Skipping experiment {experiment_number}, it is already done")

        continue

    command = f"uv run python train.py {settings} --output {model_path}"

    print(f"Running experiment {experiment_number}: {settings}")

    !{command}
```
